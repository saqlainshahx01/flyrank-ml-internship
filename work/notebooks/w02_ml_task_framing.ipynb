## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

"""
A fixed rule can't account for how much a signal like CTR should be trusted, and that depends on volume. At low impression counts (~53/90d in this data), a single click can swing CTR by ~1.9 percentage points — so the same CTR drop means something very different on a low-volume page versus a high-volume one. A model can learn that interaction between volume and rate signals; a flat if-statement threshold treats every page the same regardless of scale."""
```
